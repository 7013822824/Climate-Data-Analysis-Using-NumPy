```python
import numpy as np
import pandas as pd
```

```python
# Daily Climate time series data
data = pd.read_csv("/content/sample_data/DailyDelhiClimateTrain.csv")
data.dropna(inplace=True)
```

```python
temperature = data['meantemp'].to_numpy()
humidity = data['humidity'].to_numpy()
wind_speed = data['wind_speed'].to_numpy()
pressure = data['meanpressure'].to_numpy()
```

```python
print("Average Temperature:", np.mean(temperature))
print("Max Temperature:", np.max(temperature))
print("Min Temperature:", np.min(temperature))
print("Standard Deviation:", np.std(temperature))
```

```python
# Days with temperature > 40°C
high_temp_days = temperature > 40
print("Number of extremely hot days:", np.sum(high_temp_days))

# Days with low pressure
low_pressure_days = pressure < 1000
print("Low pressure days:", np.sum(low_pressure_days))
```

```python
fahrenheit = temperature * 9/5 + 32
print("First 5 Fahrenheit temps:", fahrenheit[:5])
```
